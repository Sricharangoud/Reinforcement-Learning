```python
import random

# Part A - Machine Initialization
machines = ['A', 'B', 'C']
epsilon = 0.2
plays = 10

# Initialize N, R and Q tables
N = {'A': 0, 'B': 0, 'C': 0}
R = {'A': 0, 'B': 0, 'C': 0}
Q = {'A': 0.0, 'B': 0.0, 'C': 0.0}

# 10 plays
for i in range(plays):
    print("\nPlay", i + 1)

    r = random.random()

    # Find highest Q machine
    max_q = max(Q.values())
    best_machines = [m for m in machines if Q[m] == max_q]
    best_machine = random.choice(best_machines)

    # Modified Epsilon-Greedy Decision
    if r < epsilon:
        print("Exploration")
        choices = [m for m in machines if m != best_machine]
        selected = random.choice(choices)
    else:
        print("Exploitation")
        selected = best_machine

    print("Selected Machine:", selected)

    # Reward Collection
    reward = float(input("Enter reward: "))

    # Update N and R
    N[selected] += 1
    R[selected] += reward

    # Incremental Q-value Update
    Q[selected] = Q[selected] + (reward - Q[selected]) / N[selected]

    print("Updated Q-value:", round(Q[selected], 2))

# Display Results
print("\n--- FINAL RESULTS ---")
print("\nMachine\tN\tR\tQ")

total_reward = 0

for m in machines:
    print(f"{m}\t{N[m]}\t{R[m]}\t{Q[m]:.2f}")
    total_reward += R[m]

print("\nTotal Reward:", total_reward)
```
