```python
import math

# Part A – Machine Initialization
machines = ['A', 'B', 'C']

# Number of plays
N_play = 10

# Part B – Initialize N, R, Q and UCB
N = {machine: 0 for machine in machines}      # Number of times played
R = {machine: 0 for machine in machines}      # Total reward
Q = {machine: 0.0 for machine in machines}    # Estimated reward
UCB = {machine: 0.0 for machine in machines}  # UCB value

total_reward = 0

print("Upper Confidence Bound (UCB) - Multi-Armed Bandit")
print("Machines:", machines)
print("Number of Plays:", N_play)
print("-" * 50)

# Part C, D and E – UCB calculation, reward collection and Q update

for play in range(1, N_play + 1):

    # Calculate UCB for every machine
    for machine in machines:

        # If machine has never been selected,
        # give it infinite UCB so it gets selected first.
        if N[machine] == 0:
            UCB[machine] = float('inf')
        else:
            UCB[machine] = Q[machine] + math.sqrt(
                (2 * math.log(play)) / N[machine]
            )

    # Select machine with highest UCB
    selected_machine = max(machines, key=lambda x: UCB[x])

    print(f"\nPlay {play}")
    print(f"Selected Machine: {selected_machine}")

    # Part D – Get reward from user
    reward = float(input("Enter reward: "))

    # Update number of times selected
    N[selected_machine] += 1

    # Update total reward
    R[selected_machine] += reward

    # Update total reward received by player
    total_reward += reward

    # Part E – Incremental sample-average Q-value update
    Q[selected_machine] = Q[selected_machine] + (
        (reward - Q[selected_machine]) / N[selected_machine]
    )

    print(f"Reward received: {reward}")
    print(f"Updated Q-value: {Q[selected_machine]:.2f}")


# Part F – Display Results

print("\n" + "=" * 70)
print("FINAL RESULTS")
print("=" * 70)

print(f"{'Machine':<10}{'N':<10}{'R':<15}{'UCB':<15}{'Q':<15}")
print("-" * 65)

# Calculate final UCB values
for machine in machines:
    if N[machine] == 0:
        UCB[machine] = float('inf')
    else:
        UCB[machine] = Q[machine] + math.sqrt(
            (2 * math.log(N_play)) / N[machine]
        )

    ucb_display = "inf" if UCB[machine] == float('inf') else f"{UCB[machine]:.4f}"

    print(
        f"{machine:<10}"
        f"{N[machine]:<10}"
        f"{R[machine]:<15.2f}"
        f"{ucb_display:<15}"
        f"{Q[machine]:<15.4f}"
    )

print("-" * 65)
print(f"Total Reward Received = {total_reward:.2f}")
```
